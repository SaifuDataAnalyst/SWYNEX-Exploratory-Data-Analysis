## 2. Dataset Overview

```python
import pandas as pd

df = pd.read_csv("student_mat_cleaned.csv", sep=",")

print("Shape:", df.shape)

print("\nColumns:", df.columns.tolist())

display(df.head())
```

## 3. Statistical Summary

```python
df.describe()
```

## 4. Categorical Variable Analysis

```python
categorical_columns = df.select_dtypes(include="str").columns

print("Categorical Columns:")
print(categorical_columns.tolist())

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts())
```

## 5. Final Grade (G3) Analysis

```python
print("G3 Statistics:")
print(df["G3"].describe())

print("\nG3 Value Counts:")
print(df["G3"].value_counts().sort_index())
```

## 6. Grade Correlation Analysis

```python
print("Correlation between grades:")

print(df[["G1", "G2", "G3"]].corr())
```

```python
import matplotlib.pyplot as plt

df[["G1", "G2", "G3"]].plot(kind="box")

plt.title("Distribution of G1, G2 and G3 Grades")
plt.ylabel("Grade")
plt.show()
```

## 7. Absences vs Final Grade

```python
print("Absences Statistics:")
print(df["absences"].describe())

print("\nCorrelation between Absences and G3:")
print(df[["absences", "G3"]].corr())
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.scatter(df["absences"], df["G3"], alpha=0.6)

plt.title("Absences vs Final Grade (G3)")
plt.xlabel("Absences")
plt.ylabel("Final Grade (G3)")
plt.show()
```

## 8. Study Time vs Final Grade

```python
print("Average G3 by Study Time:")

print(
    df.groupby("studytime")["G3"]
      .mean()
      .round(2)
)
```

```python
import matplotlib.pyplot as plt

df.groupby("studytime")["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time Category")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

## 9. Previous Failures vs Final Grade

```python
print("Average G3 by Previous Failures:")

print(
    df.groupby("failures")["G3"]
      .mean()
      .round(2)
)
```

```python
import matplotlib.pyplot as plt

df.groupby("failures")["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Previous Failures")
plt.xlabel("Number of Previous Failures")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

## 10. Gender vs Final Grade

```python
print("Average G3 by Gender:")

print(
    df.groupby("sex")["G3"]
      .mean()
      .round(2)
)
```

```python
import matplotlib.pyplot as plt

df.groupby("sex")["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

## 11. School vs Final Grade

```python
print("Average G3 by School:")

print(
    df.groupby("school")["G3"]
      .mean()
      .round(2)
)
```

## 13. Internet Access vs Final Grade

```python
print("Average G3 by Internet Access:")

print(
    df.groupby("internet")["G3"]
      .mean()
      .round(2)
)
```

```python
import matplotlib.pyplot as plt

df.groupby("internet")["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Internet Access")
plt.xlabel("Internet Access")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

## 14. Absence Group Analysis

```python
df["absence_group"] = pd.cut(
    df["absences"],
    bins=[-1, 5, 10, 20, 30, 100],
    labels=["0-5", "6-10", "11-20", "21-30", "31+"]
)

print("Average G3 by Absence Group:")

print(
    df.groupby("absence_group", observed=False)["G3"]
      .mean()
      .round(2)
)
```

```python
df.groupby("absence_group", observed=False)["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Absence Group")
plt.xlabel("Absence Group")
plt.ylabel("Average G3")
plt.xticks(rotation=0)
plt.show()
```

## 15. Correlation Matrix and Heatmap

```python
numeric_cols = df.select_dtypes(include="number")

correlation_matrix = numeric_cols.corr()

print(correlation_matrix.round(2))
```

## 16. Main EDA Visualizations

```python
plt.figure(figsize=(8, 5))

plt.hist(df["G3"], bins=10)

plt.title("Distribution of Final Grade (G3)")
plt.xlabel("Final Grade (G3)")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["G2"], df["G3"], alpha=0.6)

plt.title("G2 vs Final Grade (G3)")
plt.xlabel("G2")
plt.ylabel("G3")

plt.show()
```

```python
df.groupby("studytime")["G3"].mean().plot(kind="bar")

plt.title("Average Final Grade by Study Time")
plt.xlabel("Study Time")
plt.ylabel("Average G3")
plt.xticks(rotation=0)

plt.show()
```

## 17. EDA Key Values

```python
print("===== EDA KEY VALUES =====")

print("\n1. G3 Statistics:")
print(df["G3"].describe().round(2))

print("\n2. Average G3 by Study Time:")
print(df.groupby("studytime")["G3"].mean().round(2))

print("\n3. Average G3 by Previous Failures:")
print(df.groupby("failures")["G3"].mean().round(2))

print("\n4. Average G3 by Gender:")
print(df.groupby("sex")["G3"].mean().round(2))

print("\n5. Average G3 by School:")
print(df.groupby("school")["G3"].mean().round(2))

print("\n6. Average G3 by Internet Access:")
print(df.groupby("internet")["G3"].mean().round(2))

print("\n7. Correlation with G3:")
print(df.select_dtypes(include="number").corr()["G3"].sort_values(ascending=False).round(2))
```

## 18. Final EDA Findings

```python
print("========== FINAL EDA FINDINGS ==========")

print("\n1. Dataset Overview")
print(f"Total Students: {len(df)}")
print(f"Total Features: {df.shape[1]}")

print("\n2. Final Grade (G3)")
print(f"Average G3: {df['G3'].mean():.2f}")
print(f"Minimum G3: {df['G3'].min()}")
print(f"Maximum G3: {df['G3'].max()}")

print("\n3. Highest Average G3 by Study Time")
study_avg = df.groupby("studytime")["G3"].mean()
print(study_avg.round(2))
print("Highest:", study_avg.idxmax(), "with", round(study_avg.max(), 2))

print("\n4. Previous Failures")
failure_avg = df.groupby("failures")["G3"].mean()
print(failure_avg.round(2))
print("Highest average:", failure_avg.idxmax(), "failures")

print("\n5. Gender")
gender_avg = df.groupby("sex")["G3"].mean()
print(gender_avg.round(2))

print("\n6. School")
school_avg = df.groupby("school")["G3"].mean()
print(school_avg.round(2))

print("\n7. Internet Access")
internet_avg = df.groupby("internet")["G3"].mean()
print(internet_avg.round(2))

print("\n8. Correlation with G3")
print(
    df.select_dtypes(include="number")
      .corr()["G3"]
      .sort_values(ascending=False)
      .round(2)
)
```

## 19. EDA Summary

```python
summary = pd.DataFrame({
    "Analysis": [
        "Average Final Grade",
        "Average Grade 1",
        "Average Grade 2",
        "Average Absences",
        "Average Study Time",
        "Average Previous Failures"
    ],
    "Value": [
        df["G3"].mean(),
        df["G1"].mean(),
        df["G2"].mean(),
        df["absences"].mean(),
        df["studytime"].mean(),
        df["failures"].mean()
    ]
})

summary["Value"] = summary["Value"].round(2)

display(summary)
```
